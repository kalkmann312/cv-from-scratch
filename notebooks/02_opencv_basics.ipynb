This notebook contains basic image processing operations using OpenCV and visualization using Matplotlib.


## 1. Imports and Setup

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Visualizing a Random Image with Matplotlib

```python
img = np.random.randint(0, 256, (100, 100), dtype=np.uint8)

plt.imshow(img, cmap="gray")
plt.title("Randon generated image")
plt.axis("off")
plt.show()
```

## 3. Reading and Displaying a Real Image

```python
# Load an image from the parent directory (../images)
img = cv2.imread("../images/sax.jpg")
print("Shape:", img.shape)
```

```python
# Convert from BGR to RGB for correct color display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Image in colors")
plt.axis('off')
plt.show()
```

## 4. Resizing the Image

```python
resized = cv2.resize(img, (200, 200))  # novo tamanho (largura, altura)

plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title("Redimensioned image")
plt.axis('off')
plt.show()
```

## 5. Cropping and Rotating the Image

```python
# Crop a 800x800 region starting at (200,200)
crop = img[200:800, 200:800]
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
plt.title("Cut")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title("Rotation 90°")

plt.show()
```

## 6. Saving the Rotated Image

```python
cv2.imwrite("../images/rotated_sax.jpg", rotated)
```
