## Seção 1: Criação de Arrays

```python
import numpy as np

# 1.1 Crie um array 1D com os números de 0 a 9
a = np.arange(0, 10)
print(a)
```

```python
# 1.2 Crie um array 2D de tamanho 3x3, preenchido apenas com o número 7
b = np.full((3, 3), 7)
print(b)
```

```python
# 1.3 Gere uma matriz identidade 4x4
c = np.eye(4)
print(c)
```

## Seção 2: Indexação e Slicing

```python
# 2.1 Trabalhar com a matriz `mat`
mat = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

# a) Segunda linha
print(mat[1])
# b) Última coluna
print(mat[:, 2])
# c) Elemento central
print(mat[1, 1])

```

```python
# 2.2 Modificar valor 80 para 0
mat[2, 1] = 0
print(mat)
```

## Seção 3: Operações com Arrays

```python
# 3.1 Soma, multiplicação, média e máximo
f = np.arange(2, 10, 2)
g = np.arange(1, 9, 2)
print(f"{f} \n{g}")
print("Soma: ", f + g)
print("Multiplicação: ", f * g)
print("Média de f: ", np.mean(f))
print("Máximo de g: ", np.max(g))
```

```python
# 3.2 15 valores igualmente espaçados entre 0 e 5
espaços = np.linspace(0, 5, 15)
print(espaços)
```

## Seção 4: Desafios

```python
# 4.1 Matriz 5x5 com valores de 1 a 25
i = np.arange(1, 26).reshape(5, 5)
print("Original:", i)
print("Flip horizontal:\n", i[:, ::-1])
print("Flip vertical:\n", i[::-1, :])
```

```python
# 4.2 Soma da diagonal principal
print("Soma da diagonal: ", np.trace(i))
```

```python
# 4.3 Contar quantos valores de 1 a 50 são múltiplos de 3
arr = np.arange(1, 51)
print(arr)
print("Total de multiplos de 3: ", np.sum(arr % 3 == 0))
```
