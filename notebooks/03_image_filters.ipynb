This notebook demonstrates essential image preprocessing techniques using OpenCV and Matplotlib, such as grayscale conversion, blurring, edge detection, and thresholding.

## 1. Imports and Setup
We begin by importing the necessary libraries: OpenCV for image processing, NumPy for array handling, and Matplotlib for visualization.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load Image and Convert to Grayscale
We load a color image, convert it from BGR to RGB for visualization, and also to grayscale for processing.

```python
img = cv2.imread("../images/sax.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

## 3. Apply Gaussian Blur
Gaussian blur is used to reduce image noise and detail. It is commonly applied before edge detection to reduce false positives.


```python
blurred = cv2.GaussianBlur(gray, (9, 9), 0)
```

## 4. Detect Edges with Canny
The Canny edge detector finds edges in an image based on intensity gradients. We define two thresholds: one for strong edges and another for weak edges.

```python
edges = cv2.Canny(gray, 100, 200)
```

## 5. Apply Thresholding
Thresholding converts the grayscale image into a binary image. All pixels above a certain value become white, and those below become black.


```python
ret, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
```

```python
fig, axs = plt.subplots(1, 4, figsize=(16, 5))

axs[0].imshow(gray, cmap='gray')
axs[0].set_title("Greyscale")

axs[1].imshow(blurred, cmap='gray')
axs[1].set_title("Gaussian Blur")

axs[2].imshow(edges, cmap='gray')
axs[2].set_title("Canny Edges")

axs[3].imshow(thresh, cmap='gray')
axs[3].set_title("Threshold")

for ax in axs:
    ax.axis('off')

plt.tight_layout()
plt.show()
```
